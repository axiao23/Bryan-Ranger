```python
import os
import csv

def export_filenames_without_m(folder_path, output_file='output.csv'):
    # Open the CSV file for writing
    with open(output_file, mode='w', newline='') as file:
        writer = csv.writer(file)
        # Write header
        writer.writerow(["Filename"])

        # Iterate over all files in the directory
        for filename in os.listdir(folder_path):
            full_path = os.path.join(folder_path, filename)
            if os.path.isfile(full_path) and '_m' not in filename:
                writer.writerow([filename])

    print(f"CSV file created: {output_file}")

export_filenames_without_m("/Users/taliacho/Downloads/Bryan-Ranger/local_data/abs_neonatal_bw")
```
