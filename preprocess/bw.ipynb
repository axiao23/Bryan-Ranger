```python
from PIL import Image
import os
```

```python
def convert_to_grayscale_and_save_as_jpeg(source_folder, output_folder):
    """
    Converts all .jpg and .jpeg images in the source_folder to grayscale,
    renames .jpg to .jpeg, and saves all output as .jpeg into output_folder.
    """
    os.makedirs(output_folder, exist_ok=True)
    count = 0

    for filename in os.listdir(source_folder):
        if filename.lower().endswith((".jpg", ".jpeg")):
            input_path = os.path.join(source_folder, filename)
            base_name = os.path.splitext(filename)[0]
            output_filename = base_name + ".jpeg"
            output_path = os.path.join(output_folder, output_filename)

            try:
                with Image.open(input_path) as img:
                    grayscale_img = img.convert("L")  # Convert to grayscale
                    grayscale_img.save(output_path, "JPEG")
                    print(f"✅ Saved grayscale: {output_filename}")
                    count += 1
            except Exception as e:
                print(f"⚠️ Error processing {filename}: {e}")
    
    print(f"\n🎉 Done! {count} grayscale .jpeg images saved to: {output_folder}")
```

```python
source_folder = "/Users/taliacho/Downloads/Bryan-Ranger/clinical_data/ABS"
output_folder = "/Users/taliacho/Downloads/Bryan-Ranger/clinical_data/ABS_bw"

convert_to_grayscale_and_save_as_jpeg(source_folder, output_folder)
```
