```python
import os
import shutil
import random
from PIL import Image
import os
```

```python
def merge_and_convert_images(folder1, folder2, total_images, output_folder):
    """
    Merges total_images from two folders, converts all to grayscale,
    saves them as .jpeg in output_folder.
    """
    os.makedirs(output_folder, exist_ok=True)

    # Get all images (only .jpg or .jpeg)
    images1 = [f for f in os.listdir(folder1) if f.lower().endswith((".jpg", ".jpeg"))]
    images2 = [f for f in os.listdir(folder2) if f.lower().endswith((".jpg", ".jpeg"))]

    # Calculate how many to sample from each folder (rough split)
    split1 = total_images // 2
    split2 = total_images - split1

    sampled1 = random.sample(images1, min(split1, len(images1)))
    sampled2 = random.sample(images2, min(split2, len(images2)))

    selected_images = [(folder1, f) for f in sampled1] + [(folder2, f) for f in sampled2]
    random.shuffle(selected_images)  # Optional shuffle after merging

    count = 0
    for folder, filename in selected_images:
        input_path = os.path.join(folder, filename)
        base_name = os.path.splitext(filename)[0]
        output_filename = base_name + ".jpeg"
        output_path = os.path.join(output_folder, output_filename)

        try:
            with Image.open(input_path) as img:
                grayscale_img = img.convert("L")
                grayscale_img.save(output_path, "JPEG")
                print(f"✅ Saved grayscale: {output_filename}")
                count += 1
        except Exception as e:
            print(f"⚠️ Failed to process {filename}: {e}")

    print(f"\n🎉 Done! {count} grayscale .jpeg images saved to: {output_folder}")
```

```python
folder1 = "/Users/taliacho/Downloads/Bryan-Ranger/clinical_data/ABS_bw"
folder2 = "/Users/taliacho/Downloads/Bryan-Ranger/local_data/abs_neonatal_bw"
output_folder = "/Users/taliacho/Downloads/Bryan-Ranger/local_clinical_data/50p_50c"
total_images = 226  # Adjust as needed

merge_and_convert_images(folder1, folder2, total_images, output_folder)
```
