```python
from PIL import Image, ImageDraw
from IPython.display import display
```

```python
# Load an image from a file
image_path = "/Users/taliacho/Downloads/Ranger/Bryan-Ranger/preprocess/biceps + quads infants/capture_11_2024-08-10T17-52-33.jpeg"
image = Image.open(image_path)

# Get the size of the image
width, height = image.size
print(f"Original image size: {width}x{height}")
```

```python
# Define the corners for cropping
# Top-left corner (x1, y1)
x1, y1 = 367, 149
# Bottom-right corner (x2, y2)
x2, y2 = width - 370, height - 150

# Draw the points on the image
draw = ImageDraw.Draw(image)
point_radius = 5

# Draw a circle at the top-left corner
draw.ellipse((x1 - point_radius, y1 - point_radius, x1 + point_radius, y1 + point_radius), fill="red", outline="red")

# Draw a circle at the bottom-right corner
draw.ellipse((x2 - point_radius, y2 - point_radius, x2 + point_radius, y2 + point_radius), fill="red", outline="red")

# Display the image with points marked directly in the notebook
display(image)
```

```python
import os

def crop(directory, x1, y1, x2, y2):
    for filename in os.listdir(directory):
        if filename.endswith((".jpg", ".jpeg", ".png", ".bmp", ".tiff")):  # Add more extensions if needed
            image_path = os.path.join(directory, filename)
            with Image.open(image_path) as img:
                # Perform cropping
                cropped_img = img.crop((x1, y1, x2, y2))
                
                # Create a new filename for the cropped image
                base, ext = os.path.splitext(filename)
                cropped_filename = f"{base}_c{ext}"
                cropped_image_path = os.path.join('/Users/taliacho/Downloads/Ranger/Bryan-Ranger/local_data/bq crop', cropped_filename)
                
                # Save the cropped image
                cropped_img.save(cropped_image_path)
                print(f"Cropped image saved as {cropped_filename}")

# Example usage
# Specify the directory path and crop coordinates
directory_path = "/Users/taliacho/Downloads/Ranger/Bryan-Ranger/preprocess/biceps + quads infants"

crop(directory_path, x1, y1, x2, y2)
```
