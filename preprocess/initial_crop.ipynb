```python
import os
import cv2
import numpy as np
from PIL import Image 
```

```python
def crop(image):
    height, width = image.shape[:2]
    top = 0
    bottom = height - 50
    left = 30
    right = width - 150

    while top < bottom:
        row = image[top, :]
        non_black_pixels = np.count_nonzero(row)
        if non_black_pixels < 0.7 * width:
            break
        #increment the index to check the next row.
        top += 1

    while bottom > top:
        row = image[bottom, :]
        non_black_pixels = np.count_nonzero(row)
        if np.count_nonzero(row) > 0.4 * width:
            break
        bottom -= 1

    while left < right:
        column = image[:, left]
        #if the number of non-black pixels are greater than 70%
        if np.count_nonzero(column) > 0.8 * height:
            break
        left += 1

    while right > left:
        column = image[:, right]
        if np.count_nonzero(column) > 0.8 * height:
            break
        right -= 1

    return image[top:bottom+1, left:right+1]
```

```python
def process_files(directory):
    print(f"Processing files in directory: {directory}")
    for filename in os.listdir(directory):
        if filename.lower().endswith(".jpg") or filename.lower().endswith(".jpeg"):
            if filename.lower().startswith("cropped_"):
                continue
            file_path = os.path.join(directory, filename)
            print(f"Processing file: {file_path}")
            try:
                image = cv2.imread(file_path)
                if image is not None:
                    cropped_image = crop(image)
                    new_file_path = os.path.join('/Users/taliacho/Downloads/Ranger/Bryan-Ranger/local_data/UM', "c_" + filename)
                    cv2.imwrite(new_file_path, cropped_image)
                    print(f"Cropped and saved: {new_file_path}")
                else:
                    print(f"Failed to read image: {file_path}")
            except Exception as e:
                print(f"Error processing file {file_path}: {e}")
```

```python
def process_all(directory):
    for root, dirs, files in os.walk(directory):
        for file in files:
            if file.lower().endswith('.jpg') or file.lower().endswith('.jpeg'):
                process_files(root)
```
